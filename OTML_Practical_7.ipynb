```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2
```

```python
x = np.linspace(-10, 10, 400)

y = f(x)
```

```python
plt.figure(figsize=(8,5))

plt.plot(x, y)

plt.xlabel("x")

plt.ylabel("f(x)")

plt.title("Convex Function: f(x) = x²")

plt.grid(True)

plt.show()
```

```python
minimum_x = 0

minimum_y = f(minimum_x)

print("Global Minimum at:")

print("x =", minimum_x)

print("f(x) =", minimum_y)
```

Only one minimum exists
Gradient Descent always reaches optimal solution.

```python
plt.figure(figsize=(8,5))

plt.plot(x, y)

plt.scatter(minimum_x, minimum_y, s=100)

plt.xlabel("x")

plt.ylabel("f(x)")

plt.title("Global Minimum of Convex Function")

plt.grid(True)

plt.show()
```

The convex function has a single global minimum.
Gradient descent can reliably converge to this minimum.
Convex optimization provides stable and efficient learning in machine learning models.
